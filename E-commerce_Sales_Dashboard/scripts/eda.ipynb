## Exploraory Data Analysis

### Import Libraries & connect to postgresql

```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine('postgresql://postgres:250899@localhost:5432/Olist_Brazilian')
```

#### Pull data from postgresql

```python
## pull data from sql view
dim_date = pd.read_sql('SELECT * FROM dim_date', con=engine)
dim_customer = pd.read_sql('SELECT * FROM dim_customer', con=engine)
dim_product = pd.read_sql('SELECT * FROM dim_product', con=engine)
fact_orders = pd.read_sql('SELECT * FROM fact_orders', con=engine)
fact_payments = pd.read_sql('SELECT * FROM fact_payments', con=engine)
fact_order_items = pd.read_sql('SELECT * FROM fact_order_items', con=engine)

print("Data pulled from SQL views successfully.")
```
